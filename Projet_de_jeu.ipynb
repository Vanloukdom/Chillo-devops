```python
import random

def generate_number():
    return random.randint(1, 10)

def saisie_nombre():
    try:
        nombre = int(input("Veuillez saisir un nombre compris entre 1 et 10 \n"))
        return nombre
    except:
        print("---------------------------")
        print("Veuillez saisir un nombre\n")
        print("---------------------------")
        return None

def continuer_jouer():
    continuer = input("Voulez-vous continuer à jouer, oui/non ? ")
    if continuer.lower() not in ("oui", "non"):
        print("---------------------------")
        print("Demande non valide \n")
        print("---------------------------")
        return None
    return continuer.lower()

def main():
    print("Bienvenue dans mon jeu de GO")  # C'est pas vraiment un jeu de go, c'est juste pour la blague
    continuer = "oui"
    while continuer == "oui":
        nombre_aleatoire = generate_number()
        nombre_cherche = 0
        ptr = 0
        boucle = 5
        liste = []

        while nombre_cherche != nombre_aleatoire and ptr < 5:
            nombre_cherche = saisie_nombre()
            if nombre_cherche is None:
                continue  # Si l'utilisateur ne saisit pas un nombre, on relance la boucle
            liste.append(nombre_cherche)
            boucle -= 1
            ptr += 1

            if nombre_cherche > nombre_aleatoire:
                print("Le chiffre recherché est moins")
                print(f"Il vous reste {boucle} tentatives")
            elif nombre_cherche < nombre_aleatoire:
                print("Le chiffre recherché est plus")
                print(f"Il vous reste {boucle} tentatives")

        if nombre_cherche == nombre_aleatoire:
            print("Vous avez gagné !")
        else:
            print("Vous avez perdu, le nombre était :", nombre_aleatoire)

        continuer = continuer_jouer()
        if continuer is None:
            continuer = "non"  # Si la réponse n'est pas valide, on arrête le jeu

    print("Liste des essais :", liste)
    print("Bye")

main()
```
